```python
#import keras
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import json
import network_creation
import joblib
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
import stats_creation
```

```python
x, y = joblib.load('train_dwt_four.sav')
```

```python
EVENT_DICT = {
    'fnsz': 1,
    'gnsz': 2,
    'spsz': 3,
    'cpsz': 4,
    'absz': 5,
    'tnsz': 6,
    'cnsz': 7,
    'tcsz': 8,
    'atsz': 9,
    'mysz': 10,
    'bckg': 0
}
```

```python
pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='constant', fill_value=0.0)),
    ('scaler', MinMaxScaler()),
])
x = pipe.fit_transform(x)
```

```python
print(type(x))
#print(x.shape)
print(type(y))
```

```python
y = y.apply(lambda x: EVENT_DICT[x])
x_np = x.T
y_np = y.to_numpy().T
```
